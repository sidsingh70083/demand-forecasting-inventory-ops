```python
# ---------------------------------------------
# Phase 1: Problem Understanding & Data Setup
# Demand Forecasting Project
# ---------------------------------------------

import pandas as pd
import numpy as np

# Load dataset
data = pd.read_csv("train.csv")

# Basic inspection
print("Dataset Shape:", data.shape)
print("\nColumns:", data.columns.tolist())
print("\nFirst 5 rows:")
print(data.head())
```

```python
# Convert date to datetime
data['date'] = pd.to_datetime(data['date'])

# Sort chronologically
data = data.sort_values('date')

print("\nDate range:")
print("From:", data['date'].min())
print("To:", data['date'].max())
```

```python
print("\nMissing values:")
print(data.isnull().sum())
```

```python
store_id = 1
item_id = 1

filtered_data = data[(data['store'] == store_id) & (data['item'] == item_id)].copy()

filtered_data = filtered_data.sort_values('date')

print("\nFiltered Data Shape:", filtered_data.shape)
print(filtered_data.head())
```

```python
print("\nSummary statistics:")
print(filtered_data['sales'].describe())
```
