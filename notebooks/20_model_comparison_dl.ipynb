```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
comparison_df = pd.DataFrame({

    "Model": [
        "Naive",
        "Rolling Mean",
        "Linear Regression",
        "XGBoost",
        "ANN",
        "LSTM"
    ],

    "MAE": [
        6.29,
        4.25,
        4.57,
        4.18,
        4.16,
        4.61
    ],

    "RMSE": [
        7.72,
        5.35,
        5.72,
        5.25,
        5.14,
        5.86
    ]
})

comparison_df
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    comparison_df["Model"],
    comparison_df["MAE"]
)

plt.title(
    "Model Comparison Based on MAE"
)

plt.xlabel("Models")

plt.ylabel("MAE")

plt.grid(axis='y')

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.bar(
    comparison_df["Model"],
    comparison_df["RMSE"]
)

plt.title(
    "Model Comparison Based on RMSE"
)

plt.xlabel("Models")

plt.ylabel("RMSE")

plt.grid(axis='y')

plt.show()
```

```python
comparison_df.sort_values(
    by="MAE"
)
```
